## Step 8 - Reading a local file, and picking at random

Two new ideas in one go, and they are the ones that turn a script into something
useful: reading data from a **file** instead of typing it in the code, and using the
`random` module to make the program unpredictable.

### First, create the file

A Colab session starts empty. This cell writes a small word list next to your
notebook, the same way your repository already holds `names.txt`.


````python
# Creates words.txt in the current session

words = """launch
audience
insight
funnel
brand
signal
reach
loyalty
impact
story
value
growth
"""

with open("words.txt", "w", encoding="utf-8") as f:
    f.write(words)

print("words.txt written.")
print("Look at it: left panel, the folder icon, then words.txt")
````

### Then, read it

`open()` gives you access to the file, `.readlines()` gives you one string per line,
and each of those strings still ends with an invisible line break. `.strip()` removes it.

The `with` form closes the file for you, even if something goes wrong. Use it always.


````python
with open("words.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

# Clean each line and drop the empty ones
word_list = [line.strip() for line in lines if line.strip()]

print("Number of words:", len(word_list))
print("The list:", word_list)
print("The first one:", word_list[0])
print("The last one:", word_list[-1])
````

### Now, chance

`random` is part of Python, nothing to install. Three functions cover almost everything:

| Function | What it does |
|---|---|
| `random.choice(my_list)` | one item, taken at random |
| `random.sample(my_list, 3)` | three different items |
| `random.shuffle(my_list)` | reorders the list itself |

Run the cell several times. You get a different answer each time, which is the point.


````python
import random

print("One word at random:", random.choice(word_list))
print("Three different ones:", random.sample(word_list, 3))
````

> **Reproducibility.** `random.seed(42)` before your draws makes them identical every
> time you run the program. It sounds like it defeats the purpose, and it is exactly what
> you want when you are testing: a result you can compare. Try adding it above.

## Step 9 - Writing an output file, and getting it back

Reading is half the job. A program that produces something has to **write** it
somewhere, and you have to be able to retrieve it. This is the exact shape of the
final project: read a file, compute, write the answer.

### The two modes, and the trap

| Mode | What it does |
|---|---|
| `open(path, "w")` | **w**rite: creates the file, and **erases everything** that was in it |
| `open(path, "a")` | **a**ppend: adds at the end, keeps what was there |

`"w"` on an existing file destroys its content without asking and without warning.
It is the single most common way to lose your own work in this course. When in doubt,
write to a new name.

### Example: write, then read back to prove it


````python
# Write three lines to a file
with open("output_demo.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")
    f.write("second line\n")
    f.write("third line\n")

print("File written.")

# Never trust a write you have not read back
with open("output_demo.txt", "r", encoding="utf-8") as f:
    content = f.read()

print("--- what is actually in the file ---")
print(content)
````

> `\n` at the end of each `write()` is the line break. Forget it and your three lines
> become one long line. Try removing one and running the cell again.
